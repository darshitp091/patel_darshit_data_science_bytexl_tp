```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("manasgarg/ipl")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv("/root/.cache/kagglehub/datasets/manasgarg/ipl/versions/5/matches.csv")

print(df.head())
print("\n" + "="*50 + "\n")
print(df.info())
print("\n" + "="*50 + "\n")
print(df.describe())
```

```python
null_counts = df.isnull().sum()
print(null_counts)
print("\n" + "="*50 + "\n")

df_cleaned = df.dropna(subset=["winner", "player_of_match"])
print(df_cleaned.info())
```

```python
matches_per_season = df_cleaned["season"].value_counts().sort_index()
print(matches_per_season)
print("\n" + "="*50 + "\n")

most_wins = df_cleaned["winner"].value_counts().idxmax()
print("Team with most wins overall:", most_wins)
print("\n" + "="*50 + "\n")

venue_matches = df_cleaned[df_cleaned["venue"] == "M Chinnaswamy Stadium"]
print(venue_matches.head())
print("\n" + "="*50 + "\n")

top_players = df_cleaned["player_of_match"].value_counts().head(5)
print(top_players)
```
