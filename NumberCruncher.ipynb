```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/StudentsPerformance.csv")
```

```python
math_all   = df["math score"].values.astype(float)
read_all   = df["reading score"].values.astype(float)
write_all  = df["writing score"].values.astype(float)
```

```python
print("=" * 60)
print("TASK 1 – NumPy Array of 50 score")
print("=" * 60)
```

```python
score = math_all[:50]
random_score = np.random.randint(0, 100, size=50)
```

```python
print(f"Real score (first 50 math): {score}")
print(f"Random score (np.random):     {random_score}")
```

```python
print("\n" + "=" * 60)
print("TASK 2 – Descriptive Statistics")
print("=" * 60)
```

```python
mean_score   = np.mean(score)
median_score = np.median(score)
std_score    = np.std(score)
highest      = np.max(score)
lowest       = np.min(score)
p25          = np.percentile(score, 25)
p75          = np.percentile(score, 75)
```

```python
print(f"  Mean    : {mean_score:.2f}")
print(f"  Median  : {median_score:.2f}")
print(f"  Std Dev : {std_score:.2f}")
print(f"  Highest : {highest}")
print(f"  Lowest  : {lowest}")
print(f"  25th %ile: {p25}")
print(f"  75th %ile: {p75}")
```

```python
print("\n" + "=" * 60)
print("TASK 3 – Boolean Masking: Fails & Distinctions")
print("=" * 60)
```

```python
fail_mask        = score < 40
distinction_mask = score > 85
```

```python
fail_mask        = score < 40
distinction_mask = score > 85
```

```python
fail_score        = score[fail_mask]
distinction_score = score[distinction_mask]
```

```python
fail_indices        = np.where(fail_mask)[0]
distinction_indices = np.where(distinction_mask)[0]
```

```python
print(f"  Fail score (<40)       : {fail_score}")
print(f"  Fail student indices    : {fail_indices}")
print(f"  Total failures          : {len(fail_score)}")
print()
print(f"  Distinction score (>85): {distinction_score}")
print(f"  Distinction indices     : {distinction_indices}")
print(f"  Total distinctions      : {len(distinction_score)}")
```

```python
print("\n" + "=" * 60)
print("TASK 4 – Normalize score  [0, 1]")
print("=" * 60)
```

```python
normalized = (score - score.min()) / (score.max() - score.min())
```

```python
print(f"  Original   : {score}")
print(f"  Normalized : {np.round(normalized, 3)}")
print(f"  Min after norm: {normalized.min():.4f}  (should be 0.0)")
print(f"  Max after norm: {normalized.max():.4f}  (should be 1.0)")
```

```python
print("\n" + "=" * 60)
print("TASK 5 – Reshape  5 subjects × 10 students")
print("=" * 60)
```

```python
 matrix = score.reshape(5, 10)
 row_averages = np.mean(matrix, axis=1)
col_averages = np.mean(matrix, axis=0)
```

```python
subject_labels = ["Math", "Reading", "Writing", "Science", "Social"]
```

```python
print("  Score Matrix (5 × 10):")
print(matrix)
print()
print("  Row-wise averages (per subject):")
for label, avg in zip(subject_labels, row_averages):
    print(f"    {label:10s}: {avg:.2f}")
print()
print(f"  Col-wise averages (per student 1-10): {np.round(col_averages, 2)}")
```
